$$VIF$$

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
X=df.drop('MedHouseVal',axis=1)
vif_list=[]
arr_values=X.values
cols=len(X.columns)
for i in range(cols):
    vif_list.append(variance_inflation_factor(X.values,i))
vif_data=pd.DataFrame(vif_list,
                     index=X.columns,
                     columns=['VIF'])
vif_data
```

```python
d = vif_data.sort_values(by='VIF', ascending=False)
d
```

**$Train-Test-Split$**

```python
# Data is divided into 4 parts:

# 1. X_train, y_train  → used to train (fit) the model
# 2. X_test, y_test    → used to test (evaluate) the model

# Model is developed (trained) using X_train and y_train

# Model is tested by passing X_test, which gives predictions (y_pred)

# Finally, y_pred is compared with y_test to check accuracy/performance
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,
                                               y,
                                               test_size=0.2,
                                               random_state=1234,
                                              )
```

```python
**Assumptions**

Linearity → Relationship between X and Y is linear.

Independence → Errors are independent of each other.

Homoscedasticity → Errors have constant variance.

Normality → Errors are normally distributed.

No Multicollinearity → Predictors are not highly correlated.

No Autocorrelation → Errors are not correlated over time.
```
